```python
# Combining the useful elements from the 'metadata' and 'content' features of
# the CGIAR dataset

def get_search_text(record):

  metadata = record["metadata"]

  title = metadata.get("title") or ""
  description = metadata.get("description") or ""
  keywords = metadata.get("keywords") or []
  content = record.get("content") or ""

  if isinstance(keywords, list):
    keywords_text = " ".join(str(k) for k in keywords)

  else:
    keywords_text = str(keywords)

  return " ".join([
      title,
      description,
      keywords_text,
      content
  ]).lower()
```

Derive more 'SLM-Features' from the metadata of the CGIAR dataset for our SLM Taxonomy

```python
# Derive for 'crop'

import re

def detect_crop(text):

  '''
      This function helps to categorize input texts based on the presence of specific crop-related terms,
      aiming to assign a single, most relevant crop, or a 'general' tag if multiple or no specific
      crop-types are detected.
  '''

  # Creating a dictionary of crop keywords and their relations in respect to the custom of our SLM focus
  crop_keywords = {
      "maize" : ["maize", "corn"],
      "beans" : ["beans", "bean"],
      "cassava" : ["cassava"],
      "sorghum" : ["sorghum"],
      "groundnuts" : ["groundnut", "ground nuts", "groundnuts"],
      "peanuts" : ["peanut", "peanuts", "pea"],
      "livestock" : ["livestock", "cattle", "cow", "goat", "goats", "sheep",
                     "poultry", "chicken", "pig", "pigs"]
      }

  # Use a set to store unique matches
  unique_matches = set()

  for crop, keywords in crop_keywords.items():

    for keyword in keywords:

      # Use regex for whole-word matching to avoid partial-word matching (e.g., 'pea' in 'peach')
      pattern = r'\b' + re.escape(keyword) + r'\b'
      if re.search(pattern, text, re.IGNORECASE):
        unique_matches.add(crop)

  if len(unique_matches) == 1:
    return list(unique_matches)[0] # Return the single unique crop

  elif len(unique_matches) > 1:
    return "general"

  else:
    return "general"
```

```python
def detect_topic(text):

  '''
      This function attempts to categorize input text into one of the predefined
      agricultural topics and its relations by checking how many of the topic's keywords appear
      in the text. The topic with the most keyword matches is considered the 'best' topic.
      If no keywords are found for any topic, it returns None
  '''

  topic_keywords = {
        "crop_diseases": ["disease", "diseases", "mosaic", "rust", "blight",
                          "wilt", "virus", "fungal"],

        "pests": ["pest", "pests", "fall armyworm", "aphid", "stem borer",
                  "insect", "weevil"],

        "soil_health": ["soil health", "soil acidity", "soil fertility",
                        "erosion", "cover crop", "compost"],

        "climate_adaptation": ["climate change", "climate adaptation",
                               "climate resilience", "drought", "seasonal forecast",
                               "climate information", "heat stress"],

        "water_management": ["irrigation","rainwater", "water management",
                             "water harvesting", "drip irrigation",
                             "water conservation"],

        "fertiliser": ["fertilizer","fertiliser", "nitrogen", "phosphorus",
                       "potassium", "npk", "manure", "nutrient"],

        "livestock": ["livestock", "cattle", "goat", "sheep", "poultry", "chicken",
                      "fodder", "animal health"],

        "post_harvest": ["post harvest", "post-harvest", "storage", "stored grain",
                         "aflatoxin", "drying", "hermetic"]
    }


  # Creating a dictionary to store mapped word and amount of occurence in searched text
  scores = {}

  for topic, words in topic_keywords.items():

    topic_score = 0

    for word in words:

      # Ensure whole word matching for each keyword
      pattern = r'\b' + re.escape(word) + r'\b'
      if re.search(pattern, text, re.IGNORECASE):
        topic_score += 1
    scores[topic] = topic_score

  best_topic = max(scores, key=scores.get)

  # If no keywords are found for any topic, it returns "general".
  if scores[best_topic] == 0:
        return "general"

  return best_topic
```

```python
# Derive for agro-zone

import re

def detect_agro_zone(text):

    '''
      It's a simple rule-based classifier that looks for direct mentions of specific geographical or
      climatic descriptors to assign an agro-zone category to the text.
  '''

    agro_zone_keywords = {

        "highland": ["highland", "highlands", "high land", "high lands", "upland",
                     "uplands", "mountain", "mountains", "montane"],

        "semi_arid": ["semi-arid", "semi arid", "semiarid", "arid", "arid zone",
                      "arid region", "arid areas"],

        "sub_humid": ["sub-humid", "sub humid", "subhumid", "humid", "humid zone",
                      "humid region", "humid areas"],
    }

    detected_zones = []

    for zone, keywords in agro_zone_keywords.items():

        for keyword in keywords:

            pattern = r'\b' + re.escape(keyword) + r'\b'

            if re.search(pattern, text, re.IGNORECASE):
                detected_zones.append(zone)
                break

    # Convert the list into a single string
    if detected_zones:
        return ", ".join(detected_zones)

    # Return this when no zone is detected
    return None
```

We can take care of the missing values by refining the agro-zone detection function via expanding its classification conditions. Extended environmental specifications and evidences and crop presence-prevalence can be used to extend classification conditions in order to reduce the amount of missing values present.

```python
# Creating new agro-zone detection function

# Enriching the environmental specifications and evidences and crop presence-prevalence.

import re
import random

def detect_agroZone(text):

    # =========================================================
    # 1. EXPLICIT AGRO-ZONE EVIDENCE
    # =========================================================

    explicit_zone_keywords = {
        "highland": ["highland", "highlands", "high land", "high lands", "upland",
                     "uplands", "mountain", "mountains", "montane"],

        "semi_arid": ["semi-arid", "semi arid", "semiarid", "arid", "arid zone",
                      "arid region", "arid areas"],

        "sub_humid": ["sub-humid", "sub humid", "subhumid", "humid", "humid zone",
                      "humid region", "humid areas"],
    }


    explicit_zones = []

    for zone, keywords in explicit_zone_keywords.items():

        for keyword in keywords:

            pattern = r'\b' + re.escape(keyword) + r'\b'

            if re.search(pattern, text, re.IGNORECASE):
                explicit_zones.append(zone)
                break


    # =========================================================
    # 2. EXPLICIT EVIDENCE ALWAYS WINS
    # =========================================================

    if explicit_zones:
        return ", ".join(explicit_zones)

    # =========================================================
    # 3. ENVIRONMETAL EVIDENCE
    # =========================================================

    environmental_evidence = {

    "highland": [

        # Elevation / terrain
        "high elevation", "high elevations", "high altitude", "high altitudes",
        "elevated terrain", "elevated areas", "elevated region", "upland", "uplands",
        "upland area", "upland areas", "upland region", "upland regions", "mountainous",
        "mountainous area", "mountainous areas", "mountainous region", "mountainous regions",
        "mountain region", "mountain regions", "mountain slopes", "steep slopes",
        "steep terrain",

        # Temperature
        "cool temperatures", "cool temperature", "cool climate", "cooler temperatures",
        "low temperatures", "lower temperatures", "cold temperatures",

        # Highland-specific climatic description
        "highland climate", "highland environment"
    ],

    "semi_arid": [

        # Rainfall
        "low rainfall", "low precipitation", "limited rainfall", "limited precipitation",
        "low annual rainfall", "low annual precipitation", "erratic rainfall",
        "variable rainfall", "rainfall variability", "high rainfall variability",
        "unreliable rainfall", "unpredictable rainfall",

        # Growing / rainy season
        "short rainy season", "short wet season", "short growing season",
        "short growing period", "long dry season", "prolonged dry season",
        "extended dry season",

        # Water / moisture stress
        "water scarcity", "water stress", "water shortage", "moisture stress",
        "soil moisture stress",

        # Drought
        "drought", "drought conditions", "frequent drought", "frequent droughts",
        "recurrent drought",  "recurrent droughts", "prolonged drought", "prolonged droughts",
        "dry conditions", "dry climate"
    ],

    "sub_humid": [

        # Rainfall
        "moderate rainfall", "moderate precipitation", "adequate rainfall", "adequate precipitation",
        "reliable rainfall", "reliable precipitation", "seasonal rainfall","seasonal precipitation",

        # Growing / rainy season
        "long rainy season", "long wet season", "long growing season", "long growing period",
        "extended growing season",

        # Moisture / climate
        "moderate moisture", "moderate humidity", "moist climate", "moist conditions", "moist environment",

        # Rainfall range language
        "higher rainfall", "higher precipitation", "sufficient rainfall", "sufficient precipitation"
    ]
}

    environmental_scores = {
        "highland": 0,
        "semi_arid": 0,
        "sub_humid": 0,
    }



    for zone, terms in environmental_evidence.items():

        for term in terms:

            pattern = r'\b' + re.escape(term) + r'\b'

            if re.search(pattern, text, re.IGNORECASE):
                environmental_scores[zone] += 1


    # =========================================================
    # 3. CROP EVIDENCE
    # =========================================================

    crop_zone_mapping = {

    "maize": {
        "highland": 2, "semi_arid": 1, "sub_humid": 3
    },

    "sorghum": {
        "highland": 1, "semi_arid": 3, "sub_humid": 2
    },

    "groundnuts": {
        "highland": 1, "semi_arid": 3, "sub_humid": 2
    },

    "cassava": {
        "highland": 0, "semi_arid": 0, "sub_humid": 3
    },

    "beans": {
        "highland": 3, "semi_arid": 2, "sub_humid": 2
    }

        }


    crop_scores = {
        "highland": 0,
        "semi_arid": 0,
        "sub_humid": 0,
    }


    for crop, zone_scores in crop_zone_mapping.items():

        crop_pattern = r'\b' + re.escape(crop) + r'\b'

        if re.search(crop_pattern, text, re.IGNORECASE):

            for zone, score in zone_scores.items():
                crop_scores[zone] += score




    # =========================================================
    # 5. COMBINE CROP + ENVIRONMENTAL EVIDENCE
    # =========================================================

    combined_scores = {}

    for zone in crop_scores:

        combined_scores[zone] = (
            crop_scores[zone] +
            environmental_scores[zone]
        )


    # =========================================================
    # 6. FIND STRONGEST INFERRED ZONE
    # =========================================================

    max_score = max(combined_scores.values())

    if max_score > 0:

        strongest_zones = [
            zone
            for zone, score in combined_scores.items()
            if score == max_score
        ]

        # If there is a tie, choose one rather than None
        return random.choice(strongest_zones)


    # =========================================================
    # 7. ABSOLUTELY NO EVIDENCE
    # =========================================================

    return None
```

```python
# Combining the useful elements from the 'metadata' and 'content' features of
# the CGIAR dataset

# We bring in 'region' and 'country' features and use them as ways to extract
# and classify different agro-zones.

def get_search_Text(record):

  metadata = record["metadata"]

  title = metadata.get("title") or ""
  description = metadata.get("description") or ""
  keywords = metadata.get("keywords") or []
  content = record.get("content") or ""
  country = metadata.get("country") or ""
  region = metadata.get("region") or ""

  if isinstance(keywords, list):

    keywords_text = " ".join(str(k) for k in keywords)

  else:

    keywords_text = str(keywords)


  return " ".join([
      title,
      description,
      keywords_text,
      content,
      country,
      region,
  ]).lower()
```

```python
# Combine with all previous functions

def transform_cgiar_Record(record):

    metadata = record["metadata"]

    text_for_classification = get_search_Text(record)

    return {
        "document_id": metadata.get("gardian_id"),                # direct
        "title": metadata.get("title"),                           # direct
        "topic": detect_topic(text_for_classification),           # derived
        "crop": detect_crop(text_for_classification),             # derived
        "agro_zone": detect_agro_Zone(text_for_classification),   # derived
        "text": record.get("content"),                            # direct
        "origin": "cgiar",                                        # direct
        "source_url": metadata.get("url"),                        # direct
        "license": metadata.get("rights")                         # direct
    }
```

```python
# Define 'topic-document' prefix mapping

topic_prefix = {
    "crop_diseases" : "dis",
    "pests" : "pes",
    "soil_health" : "soi",
    "climate_adaptation" : "cli",
    "water_management" : "wat",
    "fertiliser" : "fer",
    "livestock" : "liv",
    "post_harvest" : "pos",
    "general" : "gen"
}
```

```python
# Generate IDs across the entire combined dataset

topic_counters = {}

new_document_ids = []

for index, row in combined_df.iterrows():

    topic = row['topic']

    # Get the prefix for this topic
    prefix_mapping = topic_prefix.get(topic, "gen")

    # Start the counter if this is the first occurence
    if topic not in topic_counters:
        topic_counters[topic] = 0

    # Increase the counter
    topic_counters[topic] += 1

    # Create the ID
    document_id = f'doc_{prefix_mapping}_{topic_counters[topic]:03d}'

    # Append the new ID to the list
    new_document_ids.append(document_id)

combined_df["document_id"] = new_document_ids
```

```python
import requests
import pdfplumber
import io
import re

def download_and_extract(url):
    response = requests.get(url, timeout=30)
    if response.status_code != 200:
        print(f"Failed: {url}")
        return ""
    with pdfplumber.open(io.BytesIO(response.content)) as pdf:
        pages = [page.extract_text() or "" for page in pdf.pages]
    return "\n".join(pages)

def clean_text(text):
    text = re.sub(r'\s+', ' ', text)               # collapse whitespace
    text = re.sub(r'[^\w\s.,;:()\-\/]', '', text)  # remove noise
    text = text.strip()
    return text

raw_texts = []
for doc in fao_sources:
    print(f"Downloading: {doc['title'][:50]}...")
    raw = download_and_extract(doc["url"])
    cleaned = clean_text(raw)
    raw_texts.append(cleaned)
    print(f"  → {len(cleaned)} chars extracted")
```

```python
!pip install langchain-text-splitters -q
```

```python
import pandas as pd

from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=400,      # small — good for SLM
    chunk_overlap=50,
    separators=["\n\n", "\n", ".", " "]
)

rows = []
for i, (doc, text) in enumerate(zip(fao_sources, raw_texts)):
    chunks = splitter.split_text(text)
    for j, chunk in enumerate(chunks):
        if len(chunk.strip()) < 60:
            continue
        rows.append({
            "document_id": f"fao_{i+1:03d}_{j+1:04d}",
            "title":       doc["title"],
            "topic":       doc["topic"],
            "crop":        doc["crop"],
            "agro_zone":   doc["agro_zone"],
            "text":        chunk.strip(),
            "origin":      "fao",
            "source_url":  doc["url"],
            "license":     "CC-BY-NC-SA-3.0-IGO",
        })

fao_df = pd.DataFrame(rows)
print(f"Total rows: {len(fao_df)}")
fao_df.head()
```

```python
boilerplate_keywords = [
    "non-commercial uses", "copyright", "all rights reserved",
    "reproduction", "chief publishing", "chief, publishing",
    "mention of specific companies", "expression of any opinion",
    "food and agriculture organization", "climate change and food systems resiliencei",
    "imply the expression", "authorized free", "preface",
    ".org or to the", "this volume is published", "editorial note",
    "table of contents", "list of figures", "list of tables",
    "acknowledgement", "foreword", "abbreviations"
]

def is_boilerplate(text):
    text_lower = text.lower()
    return any(kw in text_lower for kw in boilerplate_keywords)

before = len(fao_df)
fao_df = fao_df[~fao_df['text'].apply(is_boilerplate)].reset_index(drop=True)

fao_df['text'] = fao_df['text'].str.replace(r'^[\s\.,"]+', '', regex=True).str.strip()

fao_df = fao_df[fao_df['text'].str.len() >= 60].reset_index(drop=True)

print(f"Removed {before - len(fao_df)} boilerplate rows")
print(f"Final clean rows: {len(fao_df)}")

fao_df.head()
```
